```python
from atlas_algo_trading import *
from os import environ
from dotenv import load_dotenv
```

```python
load_dotenv()
user_mt5 = int(environ.get("USER_DEMO"))
password_mt5 = environ.get("PASSWORD_DEMO")
server_mt5 = environ.get("SERVER_DEMO")
```

## Create an object to handle the connection and disaply available symbols 

```python
conn = MT5(user_mt5,password_mt5,server_mt5)
conn.display_symbols(keyword="EUR",spread=30)
```

## Open a trade for EURUSD  

```python
ticket = conn.open_position(symbol="EURUSD",operation=1,lot=0.01,points=100,comment="Example Trade")
```

## Display trade and close it

```python
conn.get_positions(symbol="EURUSD")
```

```python
conn.close_position(ticket=ticket)
```
